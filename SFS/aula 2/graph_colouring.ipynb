```python
from z3 import *

Vertex = [1,2,3,4]

adj = {}
adj[1] = [2,3,4]
adj[2] = [1,4]
adj[3] = [1,4]
adj[4] = [1,2,3]

Colour = ['Red','Green','Blue']

# Declare the "matrix" of variables (dictionary of dictionaries)
x = {}
for i in Vertex:
  x[i] = {}
  for k in Colour:
    x[i][k] = Bool(str(i)+','+k)

s = Solver()

# At least one colour per vertex
for i in Vertex:
  s.add(Or([x[i][k] for k in Colour]))

# At most one colour per vertex
for i in Vertex:
  for k in Colour:
    for l in Colour:
      if k < l:
        s.add(Or(Not(x[i][k]),Not(x[i][l])))

# Adjacent vertices have different colours
for i in Vertex:
  for j in adj[i]:
    for k in Colour:
      s.add(Or(Not(x[i][k]),Not(x[j][k])))

if s.check() == sat:
  m = s.model()
  for i in Vertex:
    for k in Colour:
      if is_true(m[x[i][k]]):
        print("Vertex",i,"has colour",k)
else:
  print('Cannot colour graph!')
```

```python
# Is it true that vertex 2 and 3 always have the same colour?

s.add(Not(Or([And(x[2][c],x[3][c]) for c in Colour])))

if s.check() == unsat:
  print("Valid")
else:
  print("Invalid")
```
